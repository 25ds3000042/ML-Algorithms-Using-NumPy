```python
import numpy as np
```

```python
A = np.arange(12).reshape(3, 4)

print(A)
print("shape:", A.shape)
print("dtype:", A.dtype)
print("strides:", A.strides)

B = A[:, ::3]

print("\nB:")
print(B)
print("shape:", B.shape)
print("strides:", B.strides)

print("\nShares memory:", np.shares_memory(A, B))
```
