### Q17. Write your own `PCA` function (and find a hidden bug)
1. Write a function `PCA(X)` that takes a $(d, n)$ data matrix and returns `(eigval, eigvec)`, sorted in decreasing order of eigenvalue. Test it on the data from Q16.
2. Verify that the returned principal components are **orthonormal**, that is, $\mathbf{W}^T\mathbf{W} = \mathbf{I}$.
3. **Copy vs view:** the workshop version of `PCA` centers the data using `X -= ...`. Run the code below and check whether `X` outside the function changed. Explain why, and fix your function so the caller's data is never modified.

```python
def PCA_workshop(X):
    d, n = X.shape
    X -= X.mean(axis=1).reshape(d, 1)   # <-- in-place!
    C = X @ X.T / n
    eigval, eigvec = np.linalg.eigh(C)
    return np.flip(eigval), np.flip(eigvec, axis=1)

X = np.array([[2, 4, 6, 8],
              [1, 3, 2, 6]], dtype=float)
print("Before:", X)
PCA_workshop(X)
print("After: ", X)

# Write your own PCA function below
```
